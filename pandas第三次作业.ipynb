```python
import pandas as pd

# 复用原始数据
data = {
    "name": ["Tom", "Jerry", "Mike"],
    "score": [85, 92, 78],
    "gender": ["男", "女", "男"]
}
df = pd.DataFrame(data)

# 1. 选取两列
print("选取score、gender列：")
print(df[["score", "gender"]])

# 2. 选取指定索引行
print("\n选取索引0、2行：")
print(df.loc[[0, 2]])

# 3. 筛选score>80
print("\n筛选score>80的数据：")
print(df[df["score"] > 80])

# 4. 新增grade列
df["grade"] = df["score"].apply(lambda x: "优秀" if x >= 90 else "良好")
print("\n新增grade列：")
print(df)

# 5. 删除gender列
df_drop = df.drop(columns=["gender"])
print("\n删除gender列：")
print(df_drop)
```

```python
import pandas as pd

data = {
    "name": ["Tom", "Jerry", "Mike"],
    "score": [85, 92, 78],
    "gender": ["男", "女", "男"]
}
df = pd.DataFrame(data)

# 1. 分组求score均值、最大值
print("按gender分组，score均值&最大值：")
print(df.groupby("gender")["score"].agg(["mean", "max"]))

# 2. 分组统计人数
print("\n按gender分组统计人数：")
print(df.groupby("gender").size())
```

```python
import pandas as pd

data = {
    "name": ["Tom", "Jerry", "Mike"],
    "score": [85, 92, 78],
    "gender": ["男", "女", "男"]
}
df = pd.DataFrame(data)

# 1. 按score升序排序
print("按score升序排序：")
df_sort = df.sort_values(by="score")
print(df_sort)

# 2. 添加重复行并去重
df_dup = pd.concat([df, pd.DataFrame([["Tom", 85, "男"]], columns=df.columns)], ignore_index=True)
print("\n添加重复行后：")
print(df_dup)

df_unique = df_dup.drop_duplicates()
print("\n去重后：")
print(df_unique)
```

```python
import pandas as pd

# 1. 横向合并
df1 = pd.DataFrame({"id": [1, 2], "name": ["X", "Y"]})
df2 = pd.DataFrame({"id": [1, 2], "age": [20, 30]})
print("横向合并结果：")
print(pd.merge(df1, df2, on="id"))

# 2. 纵向拼接并重置索引
df_a = pd.DataFrame({"val": [1, 2]})
df_b = pd.DataFrame({"val": [3, 4]})
print("\n纵向拼接结果：")
print(pd.concat([df_a, df_b], ignore_index=True))
```

```python
import pandas as pd

data = {
    "name": ["Tom", "Jerry", "Mike"],
    "score": [85, 92, 78],
    "gender": ["男", "女", "男"]
}
df = pd.DataFrame(data)

# 1. score列标准化
df["score标准化"] = (df["score"] - df["score"].mean()) / df["score"].std()
print("score列标准化：")
print(df)

# 2. 透视表（中位数）
print("\n透视表（gender为行，score取中位数）：")
pivot_df = df.pivot_table(index="gender", values="score", aggfunc="median")
print(pivot_df)
```
